```python
love = ["sana", "monya", "sanaa"]
for item in love:
    if item == "sana":
        continue  # Skips the rest of this turn
    print(item)   # Runs for monya and sanaa
```

```python
l = ["1", "2", "3", "4", "5"]
for item in l:
    if item == "2":  # Added quotes to match the string
        continue
    print(item)  # Indented and removed quotes
```

```python
l = ["1", "2", "3", "4", "5"]

for item in l:
    if item == "2":
        continue
    if item == "4":
        break
    print(item)
```

```python
love = ["sana","i love you", "monya", "sanaa", "mc", "bc"]
for item in love:
    if item == "sana":
        continue
    if item == "mc":
        pass
    if item == "bc":
        pass
    if item == "monya":
        break
    print(item)  # Moved outside the IF blocks
```
