```python
#@title Using format()
name = "Alice"
age = 25
names = "rehan"
no = -45
print("My name is {} and I am {} years old.".format(name, age))
print("my bro {} has {} bitches.".format(names,no))
# Output: My name is Alice and I am 25 years old.
print("{0} is learning {1}".format("Alice", "Python"))  # Using positions
print("{name} is learning {language}".format(name="Alice", language="Python"))
```

```python
#@title string formating using :"f"
name = "Alice"
age = 25
print(f"My name is {name} and I am {age} years old.")
# Output: My name is Alice and I am 25 years old.
```

```python
pi = 3.14159
print(f"pi rounded to 2 decimal places is {pi:.2f}.")
# Output: pi rounded to 2 decimal places is 3.14.
```

```python
#@title Padding and Alignment
print(f"{'mehfuj':<10}")  # Left-align
print(f"{'harun':>15}")  # Right-align
print(f"{'jamadar':^10}")  # Center-align
```
